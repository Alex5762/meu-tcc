# Parte I — Diagnóstico e Teorização

## 1.1 Partes Interessadas e Parceiros

O projeto é direcionado a **professores de história do ensino médio** e seus **alunos**, que frequentemente encontram dificuldades em tornar o conteúdo da Segunda Guerra Mundial mais visual e quantitativo em sala de aula.

| Público | Perfil |
|---|---|
| Professores de história | Graduados em Licenciatura, 25–55 anos, ensino público e privado |
| Estudantes do Ensino Médio | 15–18 anos, sem formação técnica em dados |



## 1.2 Problemática Identificada

<div style="background-color: #fff3cd; border-left: 6px solid #e6a817; padding: 16px 20px; border-radius: 6px; margin: 10px 0; color: #5c3d00;">
  <strong>⚠️ Problema Central:</strong> O conteúdo sobre a Segunda Guerra Mundial é ensinado de forma factual e estática — datas, batalhas, nomes — sem que os estudantes consigam compreender a <strong>dimensão quantitativa e estratégica</strong> do conflito.
</div>

**Perguntas que raramente chegam ao estudante do ensino médio:**
- Quantas missões foram realizadas ao longo da guerra?
- Quais países foram mais bombardeados?
- Como a intensidade da guerra evoluiu entre 1939 e 1945?
- Quais foram os alvos mais frequentes dos ataques aéreos?

Grandes acervos históricos digitais — como o **THOR**, disponibilizado pelo governo americano — contêm respostas precisas para essas perguntas, mas são **inacessíveis para quem não trabalha com dados**.

<div style="background-color: #cfe2ff; border-left: 6px solid #0d6efd; padding: 16px 20px; border-radius: 6px; margin: 10px 0; color: #052c65;">
  <strong>❓ Pergunta Norteadora:</strong> Como transformar dados históricos brutos em conhecimento visual e acessível para o ensino de história no nível médio?
</div>

## 1.3 Justificativa

- O campo da **História Digital** demonstra que a análise quantitativa de registros históricos revela padrões impossíveis de perceber por métodos tradicionais.
- O curso de **Tecnologia da Informação** forma profissionais que devem aplicar ciência de dados a problemas reais — e o ensino de história é um problema real com impacto social direto.
- A **BNCC do Ensino Médio** prevê que o estudante use diferentes linguagens — incluindo a digital — para compreender o mundo. Visualizações de dados históricos respondem diretamente a essa demanda.
- O dataset utilizado **(THOR/NARA)** é um registro governamental oficial desclassificado, garantindo credibilidade histórica ao trabalho.

<div style="background-color: #d1e7dd; border-left: 6px solid #198754; padding: 16px 20px; border-radius: 6px; margin: 10px 0; color: #0a3622;">
  <strong>✅ Síntese:</strong> Este projeto demonstra que ciência de dados e humanidades não são campos opostos — são perspectivas complementares que, combinadas, produzem conhecimento com maior impacto social e pedagógico.
</div>

## 1.4 Objetivos a Serem Alcançados

1. **Desenvolver** um Jupyter Notebook em Python, documentado em português, que processe os dados do THOR e gere visualizações históricas.
2. **Produzir** gráficos e análises que traduzam os dados brutos em conhecimento histórico acessível para estudantes do ensino médio.
3. **Disponibilizar** o material como recurso educacional aberto, utilizável por qualquer professor de história sem custo.

## 2.6 Detalhamento Técnico do Projeto

O produto desenvolvido é um **Jupyter Notebook em Python** com 7 seções cobrindo todo o pipeline de ciência de dados:

```
Dataset THOR (.csv, 178.281 registros)
        ↓
ETL → Carregamento + Renomeação + Conversão de Tipos
        ↓
AED Básica → head / shape / groupby / min / max / SQL (pandasql)
        ↓
Distribuição → Histograma + Outliers (IQR) + KDE por ano
        ↓
Classificação → Barras (missões por país) + Pizza (tipos de alvo)
        ↓
Evolução → Gráfico de linha 1939–1945 (pico em 1944: Operação Overlord)
        ↓
Avançado → WordCloud + Regras de Associação (Apriori) + K-Means
        ↓
Notebook documentado em português com storytelling histórico
```

**Bibliotecas utilizadas:**

| Biblioteca | Finalidade |
|---|---|
| `pandas` | Manipulação e análise de dados tabulares |
| `numpy` | Operações matemáticas e estatísticas |
| `matplotlib` / `seaborn` | Visualizações estáticas |
| `plotly` | Gráficos interativos |
| `wordcloud` | Nuvem de palavras |
| `mlxtend` | Algoritmo Apriori (regras de associação) |
| `scikit-learn` | Algoritmo K-Means (clusterização) |
| `folium` | Mapas geográficos interativos |
| `pandasql` | Consultas SQL sobre DataFrames |

**Fonte dos dados:** THOR (*Theater History of Operations Reports*) — acervo desclassificado pelo *National Archives and Records Administration* (NARA), EUA.
