# Parte I — Diagnóstico e Teorização

## 1.1 Partes Interessadas e Parceiros

O projeto é direcionado a **professores de história do ensino médio** e seus **alunos**, que frequentemente encontram dificuldades em tornar o conteúdo da Segunda Guerra Mundial mais visual e quantitativo em sala de aula.

| Público | Perfil |
|---|---|
| Professores de história | Graduados em Licenciatura, 25–55 anos, ensino público e privado |
| Estudantes do Ensino Médio | 15–18 anos, sem formação técnica em dados |



## 1.2 Problemática Identificada

<div style="background-color: #fff3cd; border-left: 6px solid #e6a817; padding: 16px 20px; border-radius: 6px; margin: 10px 0; color: #5c3d00;">
  <strong>⚠️ Problema Central:</strong> O conteúdo sobre a Segunda Guerra Mundial é ensinado de forma factual e estática — datas, batalhas, nomes — sem que os estudantes consigam compreender a <strong>dimensão quantitativa e estratégica</strong> do conflito.
</div>

**Perguntas que raramente chegam ao estudante do ensino médio:**
- Quantas missões foram realizadas ao longo da guerra?
- Quais países foram mais bombardeados?
- Como a intensidade da guerra evoluiu entre 1939 e 1945?
- Quais foram os alvos mais frequentes dos ataques aéreos?

Grandes acervos históricos digitais — como o **THOR**, disponibilizado pelo governo americano — contêm respostas precisas para essas perguntas, mas são **inacessíveis para quem não trabalha com dados**.

<div style="background-color: #cfe2ff; border-left: 6px solid #0d6efd; padding: 16px 20px; border-radius: 6px; margin: 10px 0; color: #052c65;">
  <strong>❓ Pergunta Norteadora:</strong> Como transformar dados históricos brutos em conhecimento visual e acessível para o ensino de história no nível médio?
</div>

## 1.3 Justificativa

- O campo da **História Digital** demonstra que a análise quantitativa de registros históricos revela padrões impossíveis de perceber por métodos tradicionais.
- O curso de **Tecnologia da Informação** forma profissionais que devem aplicar ciência de dados a problemas reais — e o ensino de história é um problema real com impacto social direto.
- A **BNCC do Ensino Médio** prevê que o estudante use diferentes linguagens — incluindo a digital — para compreender o mundo. Visualizações de dados históricos respondem diretamente a essa demanda.
- O dataset utilizado **(THOR/NARA)** é um registro governamental oficial desclassificado, garantindo credibilidade histórica ao trabalho.

<div style="background-color: #d1e7dd; border-left: 6px solid #198754; padding: 16px 20px; border-radius: 6px; margin: 10px 0; color: #0a3622;">

## 2.6 Detalhamento Técnico do Projeto

O produto desenvolvido é um **Jupyter Notebook em Python** com 7 seções cobrindo todo o pipeline de ciência de dados:

```
Dataset THOR (.csv, 178.281 registros)
        ↓
ETL → Carregamento + Renomeação + Conversão de Tipos
        ↓
AED Básica → head / shape / groupby / min / max / SQL (pandasql)
        ↓
Distribuição → Histograma + Outliers (IQR) + KDE por ano
        ↓
Classificação → Barras (missões por país) + Pizza (tipos de alvo)
        ↓
Evolução → Gráfico de linha 1939–1945 (pico em 1944: Operação Overlord)
        ↓
Avançado → WordCloud + Regras de Associação (Apriori) + K-Means
        ↓
Notebook documentado em português com storytelling histórico
```

**Bibliotecas utilizadas:**

| Biblioteca | Finalidade |
|---|---|

| `plotly` | Gráficos interativos |

**Fonte dos dados:** THOR (*Theater History of Operations Reports*) — acervo desclassificado pelo *National Archives and Records Administration* (NARA), EUA.

### CONTEXTUALIZAÇÃO

Análise de dados sempre me pareceu interessante, mas os temas mais comuns da área — quantos produtos uma empresa vende por mês, quanto dinheiro um negócio fatura no ano — nunca me prenderam de verdade. O que mudou isso foi *Call of Duty: WWII*. O jogo é baseado em fatos reais da Segunda Guerra Mundial, e acompanhar as operações de perto — especialmente as batalhas aéreas, que no jogo têm um peso visual enorme — despertou uma curiosidade genuína sobre o conflito real por trás daquilo. Quando surgiu a oportunidade de escolher um tema para o projeto, a escolha foi imediata: analisar dados históricos das operações de bombardeio aéreo da Segunda Guerra Mundial com Python. Descobrir que existia um dataset real do governo americano com mais de 178 mil registros dessas missões foi o que confirmou que o projeto faria sentido.

### METODOLOGIA

Desenvolvi o projeto no Jupyter Notebook com Python, usando bibliotecas como pandas, seaborn, plotly, scikit-learn, wordcloud e pandasql. A etapa mais trabalhosa foi o ETL: dados em inglês técnico-militar, colunas com nulos e formatos inconsistentes que precisavam ser tratados antes de qualquer análise. Houve momentos em que fiquei horas tentando entender como determinadas partes do código funcionavam corretamente. Nesses momentos, recorri à IA como ferramenta de apoio — para explicar trechos de código, ajustar lógicas que não estavam funcionando e corrigir a parte ortográfica do trabalho. Isso agilizou bastante o desenvolvimento e me permitiu avançar sem travar em detalhes técnicos que consumiriam muito mais tempo sozinho. A maior dificuldade foi equilibrar o rigor técnico com uma linguagem acessível para qualquer leitor.

### REFLEXÃO APROFUNDADA

Esse projeto me mostrou que programação é muito mais difícil do que parece de fora, especialmente no começo. Há um ponto em que você olha para o código e não entende por onde começar. Essa é uma conclusão minha, mas os dados deixam essa dimensão muito mais visível do que qualquer livro texto.

### CONSIDERAÇÕES FINAIS

Concluído o projeto, fica claro que o que foi feito aqui é apenas o ponto de partida. A evolução natural seria transformar o notebook estático em um **dashboard interativo** — uma aplicação web onde qualquer professor pudesse, sem saber programar, filtrar os dados por país, ano ou tipo de alvo e gerar os gráficos em tempo real. Tecnologias como Streamlit ou Power BI permitiriam isso sem grande complexidade adicional.

Outra possibilidade seria ampliar a análise para conflitos que fazem parte da nossa própria história — como a atuação da Força Expedicionária Brasileira (FEB) na Itália, a Guerra do Paraguai ou até mesmo conflitos internos como a Guerra de Canudos. Criar mapas interativos e linhas do tempo comparativas sobre a história militar e civil do Brasil tornaria o material ainda mais rico e próximo da realidade dos alunos, aplicando a mesma tecnologia em dados nacionais.

Do ponto de vista da minha formação, esse projeto consolidou na prática o que a faculdade ensina na teoria. Pandas, SQL, algoritmos de IA — tudo isso saiu do slide e foi para o código resolvendo um problema real. A análise de dados deixou de ser abstrata e passou a ser uma ferramenta que eu me imagino usando profissionalmente. Mais do que isso, aprendi que programação não precisa ser só sobre sistemas corporativos e relatórios financeiros — ela pode ser usada para contar histórias, preservar memória histórica e tornar o conhecimento mais acessível para qualquer pessoa.
