## 1.1. Identificação das partes interessadas e parceiros

O projeto é direcionado a professores de história do ensino médio e seus alunos, que muitas vezes estudam a Segunda Guerra Mundial por meio de datas, nomes e acontecimentos isolados, sem contato com dados históricos organizados em gráficos e visualizações.

| Público | Perfil |
|---|---|
| Professores de história | Profissionais do ensino médio, atuantes em escolas públicas ou privadas |
| Estudantes do ensino médio | Jovens entre aproximadamente 15 e 18 anos, sem formação técnica em análise de dados |

A pertinência social do projeto está em aproximar o ensino de história das ferramentas digitais. O material produzido pode ajudar professores e estudantes a compreenderem melhor a dimensão quantitativa das operações aéreas da Segunda Guerra Mundial por meio de visualizações simples e comentadas.

## 1.2. Problemática e/ou problemas identificados

O conteúdo sobre a Segunda Guerra Mundial costuma ser trabalhado de forma factual, com foco em datas, batalhas, nomes de líderes e acontecimentos principais. Embora essa abordagem seja importante, ela nem sempre permite que os estudantes percebam a escala real do conflito.

Perguntas como quantas missões aéreas foram realizadas, quais países concentraram mais operações, quais tipos de alvos foram mais atingidos e como a intensidade da guerra mudou ao longo dos anos raramente aparecem de forma visual nas aulas.

Ao mesmo tempo, existem bases históricas públicas, como o dataset THOR (*Theater History of Operations Reports*), que reúnem milhares de registros reais sobre missões aéreas da Segunda Guerra Mundial. Porém, esses dados são pouco acessíveis para pessoas sem conhecimento em programação ou análise de dados.

Dessa forma, a problemática escolhida foi: como transformar dados históricos brutos em conhecimento visual e acessível para apoiar o ensino de história no nível médio?

## 1.3. Justificativa

Este projeto se justifica pela possibilidade de aplicar conhecimentos de Tecnologia da Informação e Ciência de Dados em uma demanda educacional real. Ao trabalhar com uma base pública e histórica, o grupo consegue desenvolver competências técnicas da disciplina e, ao mesmo tempo, produzir um material com potencial de uso no ensino de história.

A escolha do tema também se relaciona com a formação acadêmica do curso, pois envolve importação de dados, limpeza, análise exploratória, visualização, consultas SQL e aplicação de técnicas de inteligência artificial. Essas etapas demonstram, na prática, como a programação pode ser usada para interpretar problemas reais.

Além disso, a Base Nacional Comum Curricular do Ensino Médio valoriza o uso de diferentes linguagens, inclusive digitais, para compreender fenômenos sociais e históricos. Nesse sentido, transformar registros militares desclassificados em gráficos comentados contribui para tornar o estudo da Segunda Guerra Mundial mais visual, acessível e próximo da realidade dos estudantes.

## 1.4. Objetivos/resultados/efeitos a serem alcançados

## 1.5. Referencial teórico

O projeto se apoia em referenciais relacionados à história digital, à análise exploratória de dados e ao uso de tecnologias digitais na educação.

Moretti (2013), ao discutir o conceito de *distant reading*, defende que a análise de grandes conjuntos de dados pode revelar padrões que não seriam percebidos apenas pela leitura tradicional. Essa ideia se relaciona diretamente com o projeto, pois o notebook utiliza mais de 178 mil registros históricos para observar tendências das operações aéreas da Segunda Guerra Mundial.

Guldi e Armitage (2014) também destacam a importância do uso de grandes bases de dados para ampliar a pesquisa histórica. No caso deste trabalho, a base THOR permite cruzar dados militares com interpretação histórica, aproximando ciência de dados e humanidades.

Tukey (1977), referência na Análise Exploratória de Dados, defende que a exploração dos dados ajuda a descobrir padrões, comportamentos e relações antes de conclusões definitivas. Essa abordagem orienta o notebook, que utiliza estatísticas, agrupamentos e gráficos para compreender melhor a base estudada.

McKinney (2017), criador do pandas, contribui para a parte técnica do projeto ao apresentar ferramentas de manipulação de dados em Python. A biblioteca pandas foi uma das principais utilizadas para carregar, limpar e organizar os dados analisados.

Por fim, a BNCC (Brasil, 2018) reforça a importância do uso de diferentes linguagens, inclusive digitais, na formação dos estudantes. Isso justifica a proposta de transformar dados históricos em visualizações acessíveis para apoiar o ensino médio.

## 2.6. Detalhamento técnico do projeto

A solução desenvolvida é um Jupyter Notebook em Python voltado à análise de dados históricos da Segunda Guerra Mundial. O notebook utiliza o dataset THOR, uma base pública com registros de missões aéreas realizadas principalmente pelas forças Aliadas entre 1939 e 1945.

O projeto técnico foi dividido nas seguintes etapas:

- Importação da base de dados em formato CSV.
- Renomeação de colunas para português e conversão de datas.
- Análise exploratória básica com `head`, `shape`, `columns`, `groupby`, mínimo e máximo.
- Consulta SQL usando a biblioteca `pandasql`.
- Criação de gráficos de distribuição, barras, setores, densidade e evolução temporal.
- Remoção de outliers pelo método IQR.
- Criação de nuvem de palavras com os nomes dos alvos.
- Aplicação de regras de associação com Apriori.
- Aplicação de Regressão Logística para classificar a origem das operações aéreas com base em variáveis numéricas.

As principais bibliotecas utilizadas foram `pandas`, `numpy`, `matplotlib`, `seaborn`, `plotly`, `wordcloud`, `mlxtend`, `scikit-learn` e `pandasql`.

Eu tratei os metadados, traduzi os nomes das colunas de inglês para português usando snake_case e ajeitei as datas. Depois que entreguei essa base na versão 3.0, o resto do grupo assumiu o código para rodar as consultas em SQL com pandasql e aplicar o cálculo de IQR para limpar os outliers da distribuição de aviões.
