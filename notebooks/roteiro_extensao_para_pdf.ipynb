# 1. DIAGNÓSTICO E TEORIZAÇÃO


## 1.1. Identificação das partes interessadas e parceiros

O projeto é direcionado a **professores de história do ensino médio** e seus **alunos**, que frequentemente encontram dificuldades em tornar o conteúdo da Segunda Guerra Mundial mais visual e quantitativo em sala de aula.

| Público | Perfil |
|---|---|
| Professors de história | Graduados em Licenciatura, 25–55 anos, atuantes no ensino público e privado |
| Estudantes do Ensino Médio | 15–18 anos, sem formação técnica em análise de dados |

A pertinência social do projeto reside na democratização do conhecimento histórico: ao transformar dados brutos e inacessíveis em visualizações compreensíveis, o trabalho cria uma ponte direta entre a tecnologia da informação e o ensino de história. A demanda foi identificada a partir de conversas informais com colegas do curso e da observação de que acervos históricos digitais de alta relevância — como o dataset THOR — são praticamente desconhecidos do público não especializado em dados.


## 1.2. Problemática e/ou problemas identificados

**Problema Central:** O conteúdo sobre a Segunda Guerra Mundial é ensinado de forma factual e estática — datas, batalhas, nomes — sem que os estudantes consigam compreender a dimensão quantitativa e estratégica do conflito.

Perguntas que raramente chegam ao estudante do ensino médio:

- Quantas missões aéreas foram realizadas ao longo da guerra?
- Quais países foram mais bombardeados?
- Como a intensidade dos bombardeios evoluiu entre 1939 e 1945?
- Quais foram os alvos mais frequentes dos ataques aéreos?

Grandes acervos históricos digitais — como o **THOR** (*Theater History of Operations Reports*), disponibilizado pelo governo americano via NARA — contêm respostas precisas para essas perguntas, mas são **inacessíveis para quem não possui formação em análise de dados**.

**Pergunta Norteadora:** Como transformar dados históricos brutos em conhecimento visual e acessível para o ensino de história no nível médio?


## 1.3. Justificativa

A elaboração deste projeto justifica-se por quatro razões interdependentes:

- O campo da **História Digital** demonstra que a análise quantitativa de registros históricos revela padrões impossíveis de perceber por métodos tradicionais, conforme apontam Moretti (2013) e Guldi & Armitage (2014).

- O curso de **Tecnologia da Informação** forma profissionais que devem aplicar ciência de dados a problemas reais, com impacto social direto — e o ensino de história é um desses problemas reais.

- A **Base Nacional Comum Curricular (BNCC) do Ensino Médio** prevê que o estudante utilize diferentes linguagens — incluindo a digital — para compreender o mundo. Visualizações de dados históricos respondem diretamente a essa demanda curricular.

- O dataset utilizado (**THOR/NARA**) é um registro governamental oficial desclassificado, gerando credibilidade histórica e científica ao trabalho.

Este projeto demonstra que ciência de dados e humanidades não são campos opostos — são perspectivas complementares que, combinadas, produzem conhecimento com maior impacto social e pedagógico.

## 1.5. Referencial teórico (subsídio teórico para propositura de ações da extensão)

O projeto apoia-se em três eixos teóricos: epistemológico (História Digital), técnico (ciência de dados) e pedagógico (currículo do ensino médio).

O dataset THOR, desclassificado pelo governo americano, é exatamente o tipo de fonte primária digital que os autores recomendam para novas interpretações históricas.

**John W. Tukey**, em *Exploratory Data Analysis* (1977), estabeleceu os fundamentos da Análise Exploratória de Dados (AED), metodologia central adotada nas seções 2 a 6 do notebook desenvolvido.

O **MEC**, na *Base Nacional Comum Curricular — Ensino Médio* (2018), fundamenta a dimensão pedagógica do projeto ao estabelecer a competência digital como parte do currículo básico, validando o uso de visualizações de dados como ferramenta de ensino de história.

## 2.6. Detalhamento técnico do projeto

O produto desenvolvido é um **Jupyter Notebook em Python** estruturado em 7 seções, cobrindo o pipeline completo de ciência de dados:

**Seção 1 — ETL (Extração, Transformação e Carga)**
Carregamento do dataset THOR (178.281 registros, formato .csv), renomeação de 11 colunas para português, conversão de tipos e documentação da fonte primária.

**Seção 2 — Análise Exploratória de Dados (AED) Básica**
Inspeção inicial com `head()`, `shape`, `columns`; contagem de missões por país com `groupby`; identificação de valores mínimos e máximos; consulta SQL via pandasql.

**Seção 3 — Histograma e Remoção de Outliers**
Distribuição da quantidade de aeronaves por missão; identificação e remoção de outliers pelo método IQR (limite: 27 aeronaves por missão).

**Seção 4 — Densidade (KDE) por Ano**
Estimativa de densidade kernel mostrando a evolução da intensidade das operações entre 1942 e 1945.

**Seção 5 — Classificação: Gráficos de Barras e Setores**
Total de missões por país aliado (barras); distribuição percentual dos tipos de alvo bombardeado (pizza/setores).

**Seção 6 — Evolução Temporal**
Gráfico de linha mostrando o total de missões por ano (1939–1945), com storytelling destacando o pico de 1944 associado à Operação Overlord (Dia D).

**Seção 7 — Análises Avançadas com Inteligência Artificial**
- *Nuvem de Palavras:* cidades mais bombardeadas em destaque (Berlim, Hamburgo, Colônia).
- *Regras de Associação (Apriori):* 6 regras identificadas entre país, tipo de missão e tipo de alvo.
- *Clusterização (K-Means):* 3 clusters de missões identificados por intensidade e perfil operacional.
